```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# 1. Read the input image
A = cv2.imread('/content/drive/MyDrive/COMPUTER VISION/download.jpg', 0)

if A is None:
    print("Error: Image not found.")

else:
    # 2. Create a copy of the original image
    noisy_img = np.copy(A)

    # 3. Apply Median Filter
    median_filter = cv2.medianBlur(noisy_img, 3)

    # 4. Apply Gaussian Filter
    gaussian_filter = cv2.GaussianBlur(noisy_img, (3, 3), 0)

    # 5. Display the results
    plt.figure(figsize=(15, 5))

    # Original Image
    plt.subplot(1, 3, 1)
    plt.imshow(noisy_img, cmap='gray')
    plt.title("Original Input")
    plt.axis('off')

    # Median Filter
    plt.subplot(1, 3, 2)
    plt.imshow(median_filter, cmap='gray')
    plt.title("Median Filter")
    plt.axis('off')

    # Gaussian Filter
    plt.subplot(1, 3, 3)
    plt.imshow(gaussian_filter, cmap='gray')
    plt.title("Gaussian Filter")
    plt.axis('off')

    # 6. Adjust layout and display
    plt.tight_layout()
    plt.show()
```
